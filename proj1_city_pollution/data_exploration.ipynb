```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
```

# Heat map of pollution

```python
train_x = np.loadtxt('train_x.csv', delimiter=',', skiprows=1)
train_y = np.loadtxt('train_y.csv', delimiter=',', skiprows=1)
test_x = np.loadtxt('test_x.csv', delimiter=',', skiprows=1)
```

```python
train_y
```

```python
plt.hist(train_y[train_x[:,2] == 1],bins=100)
plt.show()
```

This is a heatmap of the pollution throughout the whole city.

```python
# Create scatter heatmap
plt.figure(figsize=(12, 8))
scatter = plt.scatter(train_x[:,0], train_x[:,1], c=train_y, 
                     cmap='plasma', s=10, alpha=0.7)

# Add colorbar
cbar = plt.colorbar(scatter)
cbar.set_label('Pollution Level')

# Add labels and title
plt.xlabel('Longitude')
plt.ylabel('Latitude')
plt.title('Pollution Level Heatmap (Scatter)')
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

This is a heatmap of the pollution levels in the candidate residential areas

```python
res_coords = train_x[train_x[:,2] == 1]
res_pollution = train_y[train_x[:,2] == 1]
```

```python
# Create scatter heatmap
plt.figure(figsize=(12, 8))
scatter = plt.scatter(res_coords[:,0], res_coords[:,1], c=res_pollution, 
                     cmap='plasma', s=10, alpha=0.7)

# Add colorbar
cbar = plt.colorbar(scatter)
cbar.set_label('Pollution Level')

# Add labels and title
plt.xlabel('Longitude')
plt.ylabel('Latitude')
plt.title('Pollution Level Heatmap (Scatter)')
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
# Create scatter heatmap
plt.figure(figsize=(12, 8))
scatter = plt.scatter(test_x[:,0], test_x[:,1], c=test_x[:,2], 
                     cmap='seismic', s=10, alpha=0.7)

# Add colorbar
cbar = plt.colorbar(scatter)
cbar.set_label('Pollution Level')

# Add labels and title
plt.xlabel('Longitude')
plt.ylabel('Latitude')
plt.title('Pollution Level Heatmap (Scatter)')
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
subsample_x, _, subsample_y, _ = train_test_split(train_x, train_y, test_size=.75, random_state=0)
np.size(subsample_x)
```

```python
# Create scatter heatmap
plt.figure(figsize=(12, 8))
scatter = plt.scatter(subsample_x[:,0], subsample_x[:,1], c=subsample_y, 
                     cmap='plasma', s=10, alpha=0.7)

# Add colorbar
cbar = plt.colorbar(scatter)
cbar.set_label('Pollution Level')

# Add labels and title
plt.xlabel('Longitude')
plt.ylabel('Latitude')
plt.title('Pollution Level Heatmap (Scatter)')
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```
